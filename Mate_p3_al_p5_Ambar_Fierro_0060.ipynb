#### problema 3
Problema 3: Un número al que se le resta 7 da como resultado 18Paso 1: Explicación y resoluciónPlanteamiento: Llamaremos $x$ al número desconocido.Ecuación: "Un número ($x$) al que se le resta 7 es igual a 18":$$x - 7 = 18$$Despeje: El 7 está restando, por lo que pasa al otro lado sumando para dejar a la $x$ sola:$$x = 18 + 7$$Solución:$$x = 25$$

```python
# Problema 3: Hallar x tal que x - 7 = 18
# resultado = 18
# resta - 7
resultado = int(input ("Ingrese el resultado: "))
resta = int(input ("Ingrese la resta: "))
# Despeje: x - 18 + 7
numero_buscado = resultado + resta
print(f"El número es: {numero_buscado}")
```

#### Problema 4
El triple de un número más 4 es igual a 31Paso 1: Explicación y resoluciónPlanteamiento:El número es $x$."El triple de un número" es $3x$.Ecuación:$$3x + 4 = 31$$Despeje paso a paso:Primero quitamos el 4 que está sumando (pasa restando):$$3x = 31 - 4$$$$3x = 27$$Luego, el 3 que multiplica a la $x$ pasa dividiendo:$$x = \frac{27}{3}$$Solución:$$x = 9$$

```python
# Problema 4:3x + 4 = 31
# total = 31
# termino_independiente = 4
# coeficiente = 3
total = int(input("Ingrese el total: "))
termino_independiente = int(input ("Ingrese el termino independiente: "))
coeficiente = int(input ("Ingrese el coeficiente: "))

# Paso a paso: (31 - 4) /3
x = (total - termino_independiente) / coeficiente

print (f"El número resultante es: {int(x)}")
```

#### Problema 5
Dimensiones de un terreno (Perímetro)Paso 1: Explicación y resoluciónPlanteamiento:Ancho del terreno: $a$Largo del terreno: $a + 5$ (mide 5 metros más que el ancho).Ecuación: El perímetro de un rectángulo es la suma de todos sus lados: $Largo + Ancho + Largo + Ancho = Perímetro$.$$(a + 5) + a + (a + 5) + a = 50$$Simplificación: Agrupamos las $a$ y los números:$$4a + 10 = 50$$Despeje:Pasamos el 10 restando:$$4a = 50 - 10$$$$4a = 40$$Pasamos el 4 dividiendo:$$a = \frac{40}{4} = 10$$Solución:Ancho: 10 metros.Largo: $10 + 5 = 15$ metros.Comprobación: $10 + 15 + 10 + 15 = 50$.

```python
#Problema 5: Perímetro = 2(largo) + 2(ancho) = 50
# perimetro_total = 50
# diferencia_largo = 5
perimetro_total = int(input ("Ingrese el perímetro total: "))
diferencia_largo = int(input("Ingrese la diferencia entre el largo y el ancho: "))

# Simplificando la ecuación 4x + 10 = 50 -> x = (50 - 10) / 4
ancho = (perimetro_total - (2 * diferencia_largo)) / 4
largo = ancho + diferencia_largo
print(f"El ancho mide: {ancho} metros")
print(f"El largo mide: {largo} metros")
```
