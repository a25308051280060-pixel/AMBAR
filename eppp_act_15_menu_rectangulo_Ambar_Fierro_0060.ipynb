Se solicita desarrollar un programa en Python que permita calcular el área y el perímetro de un rectángulo de forma interactiva. El programa deberá contar con un menú de opciones controlado por un ciclo while, utilizando estructuras condicionales if/elif/else para dirigir la ejecución hacia las funciones correspondientes. Cada función solicitará al usuario las dimensiones necesarias (base y altura) en el momento exacto de su llamada. El código se mantendrá en un enfoque puramente procedimental, sin validación de datos ni programación orientada a objetos, y se decorará con emojis para ofrecer una interfaz visual clara y motivadora, ideal para consolidar los fundamentos de control de flujo y modularización en Python.

1. Área del Rectángulo A=b⋅h
2. Perímetro del Rectángulo P=2⋅(b+h)

* Nombres de funciones: Python requiere que los identificadores sean continuos. Se usó area_rectangulo() en lugar de area rectángulo () para evitar errores de sintaxis.
* Estructura while True:: Crea un bucle infinito que solo se detiene cuando el usuario ingresa "3" y se ejecuta break.
* Condicionales: if/elif/else evalúa la cadena de texto ingresada y llama a la función correspondiente.
* Solicitud de datos: Como se indicó, input() para base y altura está dentro de cada función, por lo que el programa pedirá los datos cada vez que elijas una opción.
* Sin validación: Si escribes letras en lugar de números, Python mostrará un ValueError. Esto es intencional para mantener el código enfocado en estructuras de control y modularidad.

```python
# =============================================================================
# ⚠️ PROGRAMA DE CALCULOS GEOMETRICO
# Elaborado por Ambar Fierro
# Área del Rectángulo A=b⋅h
# Perímetro del Rectángulo P=2⋅(b+h)

import math

def area_rectangulo():
    """Calcula y muestra el área de un rectángulo."""
    print("\n🟦 CÁLCULO DEL ÁREA DEL RECTÁNGULO")
    base = float(input("📏 Ingresa la base: "))
    altura = float(input("📏 Ingresa la altura: "))
    area = base * altura
    print(f"✅ Área: {area:.2f} unidades²\n")

def perimetro_rectangulo():
    """Calcula y muestra el perímetro de un rectángulo."""
    print("\n🟩 CÁLCULO DEL PERÍMETRO DEL RECTÁNGULO")
    base = float(input("📏 Ingresa la base: "))
    altura = float(input("📏 Ingresa la altura: "))
    perimetro = 2 * (base + altura)
    print(f"✅ Perímetro: {perimetro:.2f} unidades\n")

def menu():
    """Muestra el menú principal y controla el flujo del programa."""
    while True:
        print("📐 === MENÚ DE GEOMETRÍA PLANA === 📐")
        print("1️⃣ Calcular área del rectángulo")
        print("2️⃣ Calcular perímetro del rectángulo")
        print("3️⃣ Salir")
        opcion = input("👉 Selecciona una opción (1-3): ")

        if opcion == "1":
            area_rectangulo()
        elif opcion == "2":
            perimetro_rectangulo()
        elif opcion == "3":
            print("\n👋 ¡Hasta pronto! Gracias por usar el programa. ✨")
            break
        else:
            print("\n⚠️ Opción no válida. Inténtalo de nuevo.\n")

# Punto de entrada del programa
if __name__ == "__main__":
    menu()
```
