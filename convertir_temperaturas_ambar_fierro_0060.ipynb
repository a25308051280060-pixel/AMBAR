## 1. Fórmulas de Conversión

Para convertir entre grados Celsius ($^\circ C$) y Fahrenheit ($^\circ F$), utilizamos las siguientes ecuaciones matemáticas:

* **De Celsius a Fahrenheit:**
    $$^\circ F = (^\circ C \times 1.8) + 32$$
    *(También se puede expresar como $^\circ F = (^\circ C \times \frac{9}{5}) + 32$)*

* **De Fahrenheit a Celsius:**
    $$^\circ C = \frac{^\circ F - 32}{1.8}$$
    *(También se puede expresar como $^\circ C = (^\circ F - 32) \times \frac{5}{9}$)*

## 2. Ejemplos Resueltos Paso a Paso

### Ejemplo 1: Convertir $25^\circ C$ (Temperatura ambiente) a $^\circ F$
1.  **Identificar datos:** Tenemos $25^\circ C$.
2.  **Sustituir en la fórmula:** $^\circ F = (25 \times 1.8) + 32$.
3.  **Multiplicar:** $25 \times 1.8 = 45$.
4.  **Sumar:** $45 + 32 = 77$.
5.  **Resultado:** $25^\circ C$ equivale a **$77^\circ F$**.

### Ejemplo 2: Convertir $98.6^\circ F$ (Temperatura corporal) a $^\circ C$
1.  **Identificar datos:** Tenemos $98.6^\circ F$.
2.  **Sustituir en la fórmula:** $^\circ C = \frac{98.6 - 32}{1.8}$.
3.  **Restar primero:** $98.6 - 32 = 66.6$.
4.  **Dividir:** $66.6 / 1.8 = 37$.
5.  **Resultado:** $98.6^\circ F$ equivale a **$37^\circ C$**.

## 3. Código en Python

```python
# Función para convertir de Celsius a Fahrenheit
def convertirCF():
    celsius = float(input("Ingresa los grados Celsius: "))

def convertirFC():
    fahrenheit = float(input("Ingresa los grados Fahrenheit: "))
```

* **`float()`:** Permite que el usuario ingrese números con decimales (como 36.5).
* **`f"{valor:.2f}"`:** Se usa para mostrar solo 2 decimales en el resultado final y que se vea más limpio.
* **`break`:** Rompe el bucle inmediatamente cuando eliges la opción 3.
